1.- Orden de los números

- Lee tres números por teclado.

- Evalúa si los números están en orden ascendente o no.

- Si el primer número introducido es 0, muestra un mensaje de error.
Nota: No uses listas en este ejercicio.

```python
# Seleccionamos nuestros tres numeros primero
n1 = int(input("Introduce el primer número: "))
n2 = int(input("Introduce el segundo número: "))
n3 = int(input("Introduce el tercer número: "))

# marcamos primero la condicion de que 0 es error. Y luego verificamos si los numeros estan en orden ascendente o no...
if n1 == 0:
    print("Error: el primer número no puede ser 0.")
elif n1 < n2 < n3:
    print("Los números están en orden ascendente.")
else:
    print("Los números NO están en orden ascendente.")
```

2.- Orden numérico con listas

- Repite el ejercicio anterior, pero introduciendo los números en una lista.

```python
# Seleccionamos nuestros tres numeros primero y los guardamos en una lista.
numeros = []
numeros.append(int(input("Introduce el primer número: ")))
numeros.append(int(input("Introduce el segundo número: ")))
numeros.append(int(input("Introduce el tercer número: ")))

# Comprobamos el error primero
if numeros[0] == 0:
    print("Error: el primer número no puede ser 0.")
elif numeros[0] < numeros[1] < numeros[2]:
    print("Los números están en orden ascendente.")
else:
    print("Los números NO están en orden ascendente.")
```

3.- Sumatorio de los primeros N números

- Realiza un programa que sume los primeros N números empezando en 0.

- N será un número límite introducido por el usuario.
Ejemplo: con N = 5 → sumatorio = 0 + 1 + 2 + 3 + 4

```python
n = int(input("Introduce el número límite N: "))

sumatorio = 0

for i in range(n):
    sumatorio += i

print(f"El sumatorio de los primeros {n} números es: {sumatorio}")
```

4.- Contador de letras "a"

- Lee letras una por una y cuenta cuántas veces se introduce la letra 'a'.

- El programa termina cuando el usuario introduce el carácter '.'.

- Al finalizar, muestra el número total de veces que se ha introducido la letra 'a'.

```python
contador = 0

while True:
    letra = input("Introduce una letra (. para terminar): ")

    if letra == ".":
        break
    elif letra == "a":
        contador += 1

print(f"La letra 'a' se ha introducido {contador} veces.")
```
