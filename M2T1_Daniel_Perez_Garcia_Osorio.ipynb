3.-Extraer partes de una dirección (slicing)
Con direccion = "Calle Falsa, 123, Springfield":

-Extrae e imprime solo el nombre de la calle.

-Extrae e imprime solo el número.

-Extrae e imprime solo la ciudad.

```python
direccion = "Calle Falsa, 123, Springfield"

# Nombre de la calle
calle = direccion[0:11]
print(calle)       

# Número
numero = direccion[13:16]
print(numero)      

# Ciudad
ciudad = direccion[18:]
print(ciudad)
```

4.-Identificador único con slicing y concatenación

-Define nombre, apellido y año_nacimiento (string).

-Extrae las 3 primeras letras de nombre y apellido, y los 2 últimos dígitos del año.

-Concatena para formar un identificador (p. ej. CriRod89) e imprímelo.

```python
nombre = "Cristina"
apellido = "Rodriguez"
año_nacimiento = "1989"

# 3 primeras letras de nombre y apellido
parte_nombre = nombre[0:3]
parte_apellido = apellido[0:3]

# 2 últimos dígitos del año
parte_año = año_nacimiento[-2:]

# Concatenación para formar el identificador
identificador = parte_nombre + parte_apellido + parte_año
print(identificador)   
```

6.-Analizador de palabras

-Dados: texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas." y objetivo = "Python".

-Convierte ambos a minúsculas para búsqueda insensible a mayúsculas.

-Cuenta apariciones como palabra exacta (con split()) y como subcadena (con .count()).

-Muestra ambos totales.

-Extra: genera una versión del texto remarcando las apariciones de la palabra objetivo en MAYÚSCULAS.
Salida esperada (aproximada): totales de coincidencias y texto resaltado, por ejemplo: PYTHON es POTENTE...

```python
texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas."
objetivo = "Python"

# Convertir a minúsculas
texto_lower = texto.lower()
objetivo_lower = objetivo.lower()

# Como palabra exacta (usando split, limpiando puntuación básica si es necesario, o tal cual split)
# Veamos: palabras de texto_lower divididas por espacios (o quitando puntos/comas)
import re
palabras = re.findall(r'\b\w+\b', texto_lower)
exactas = palabras.count(objetivo_lower)

# Como subcadena
subcadena_count = texto_lower.count(objetivo_lower)

# Extra: Versión remarcando las apariciones de la palabra objetivo en MAYÚSCULAS en el texto original
# Vamos a buscar todas las apariciones de 'python' (ignorando mayús/minús) en el texto original y ponerlas en mayúsculas.
# Reemplazo insensitivo manteniendo el formato original o poniendo PYTHON.
texto_resaltado = re.sub(re.escape(objetivo), "PYTHON", texto, flags=re.IGNORECASE)

print(f"Exactas: {exactas}")
print(f"Subcadena: {subcadena_count}")
print(f"Resaltado: {texto_resaltado}")
```
