3.-Extraer partes de una dirección (slicing)
Con direccion = "Calle Falsa, 123, Springfield":

-Extrae e imprime solo el nombre de la calle.

-Extrae e imprime solo el número.

-Extrae e imprime solo la ciudad.

```python
direccion = "Calle Falsa, 123, Springfield"

# Nombre de la calle
calle = direccion[0:11]
print(calle)       

# Número
numero = direccion[13:16]
print(numero)      

# Ciudad
ciudad = direccion[18:]
print(ciudad)
```

4.-Identificador único con slicing y concatenación

-Define nombre, apellido y año_nacimiento (string).

-Extrae las 3 primeras letras de nombre y apellido, y los 2 últimos dígitos del año.

-Concatena para formar un identificador (p. ej. CriRod89) e imprímelo.

```python
nombre = "Cristina"
apellido = "Rodriguez"
año_nacimiento = "1989"

# 3 primeras letras de nombre y apellido
parte_nombre = nombre[0:3]
parte_apellido = apellido[0:3]

# 2 últimos dígitos del año
parte_año = año_nacimiento[-2:]

# Concatenación para formar el identificador
identificador = parte_nombre + parte_apellido + parte_año
print(identificador)   
```

6.-Analizador de palabras

1-Datos: texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas." y objetivo = "Python".

2-Convierte ambos a minúsculas para búsqueda insensible a mayúsculas.

3-Cuenta apariciones como palabra exacta (con split()) y como subcadena (con .count()).

4-Muestra ambos totales.

5-Extra: genera una versión del texto remarcando las apariciones de la palabra objetivo en MAYÚSCULAS.
Salida esperada (aproximada): totales de coincidencias y texto resaltado, por ejemplo: PYTHON es POTENTE...

```python

# 1.Datos de texto y palabra objetivo
texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas."
objetivo = "Python"

# 2. Convertir ambos a minúsculas para búsqueda insensible a mayúsculas
texto_minusculas = texto.lower()
objetivo_minusculas = objetivo.lower()
print (texto_minusculas)
print (objetivo_minusculas)

# 3. Usar split() para contar apariciones exactas de la palabra
palabras = texto_minusculas.split(objetivo_minusculas)
conteo_split = len(palabras)

# Como subcadena (usando .count())
conteo_substring = texto_minusculas.count(objetivo_minusculas)

# 4. Muestra ambos totales
print(f"Total como palabra exacta: {conteo_split}")
print(f"Total como subcadena: {conteo_substring}")


# 5. Extra: Generar versión remarcando las apariciones en MAYÚSCULAS
# Reemplazamos la palabra objetivo (en cualquiera de sus formas de mayúsculas/minúsculas)
# usando un reemplazo básico o expresiones regulares.
import re

# Usamos regex para reemplazar la palabra ignorando mayúsculas/minúsculas
patron = re.compile(re.escape(objetivo), re.IGNORECASE)
texto_resaltado = patron.sub(objetivo.upper(), texto)

print(f"Texto resaltado: {texto_resaltado}")
```
