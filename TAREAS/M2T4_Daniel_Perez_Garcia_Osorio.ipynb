5.- Quinto bloque — Claves de entrada de usuario

 Código a evaluar:
datos = {
  "nombre": "Juan",
  "edad": 30,
  "ciudad": "Madrid"
}
clave = input("Introduce una clave para obtener su valor: ")

print(f"El valor asociado a '{clave}' es: {datos[clave]}")
                    
- Analiza el código e identifica las posibles excepciones (p. ej. KeyError).

- Implementa un control de errores que evite el fallo y muestre un mensaje claro indicando cómo introducir una clave válida.

```python
datos = {
    "nombre": "Juan",
    "edad": 30,
    "ciudad": "Madrid"
}

entrada_usuario = input("Introduce una clave para obtener su valor: ")

clave = entrada_usuario.strip().lower()

try:
    valor = datos[clave]
    print(f"El valor asociado a '{clave}' es: {valor}")

except KeyError:
    print(f"\nError: La clave '{entrada_usuario}' no es válida.")
    print("Para introducir una clave válida, asegúrate de elegir una de las siguientes opciones:")
    for k in datos.keys():
        print(f" - {k}")
```

6.- Sexto bloque — Conversión y división

Código a evaluar:
def calcular(numero1, numero2):
    return (int(numero1) ** 2) / int(numero2)

n1 = input("Introduce el primer número: ")

n2 = input("Introduce el segundo número: ")

resultado = calcular(n1, n2)

print(f"El resultado del cálculo es: {resultado}")
                    
- Analiza el código e identifica las excepciones que pueden ocurrir (p. ej. ValueError, ZeroDivisionError).

- Implementa un control con try-except que evite el fallo y explique al usuario cómo corregir la entrada.

```python
def calcular(numero1, numero2):
    return (int(numero1) ** 2) / int(numero2)

n1 = input("Introduce el primer número: ")
n2 = input("Introduce el segundo número: ")

try:
    resultado = calcular(n1, n2)
    print(f"El resultado es: {resultado}")

except ValueError:
    print("\nError: Una o ambas entradas no son números enteros válidos.")
    print("Cómo corregirlo:Introduzca únicamente números enteros.")

except ZeroDivisionError:
    print("\nError: No es posible realizar una división entre cero.")
    print("Cómo corregirlo: El segundo número debe ser diferente de 0.")
```

7.- Séptimo bloque — Función para agregar sin repetir

- Define una función llamada agregar_sin_repetidos() que reciba una lista y un elemento.

- La función debe añadir el elemento al final de la lista siempre que no exista ya en ella.

- Si el elemento ya se encuentra en la lista, debe lanzar un ValueError que se capture mostrando el mensaje:

- Error: Imposible añadir elementos duplicados => [elemento]

```python
def agregar_sin_repetidos(lista, elemento):
    if elemento in lista:
        raise ValueError(f"Error: Imposible añadir elementos duplicados => {elemento}")
    lista.append(elemento)

# Ejemplo de prueba
mi_lista = [10, 20, 30]

"""Intento 1"""
try:
    agregar_sin_repetidos(mi_lista, 40)
    print("Lista actualizada:", mi_lista)
except ValueError as e:
    print(e)

"""Segundo intento"""
try:
    agregar_sin_repetidos(mi_lista, 20)
    print("Lista actualizada:", mi_lista)
except ValueError as e:
    print(e)
```
