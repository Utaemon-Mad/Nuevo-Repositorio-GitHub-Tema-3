1.- Primer bloque — División por cero

 Código a evaluar:
numero = 7/0
                    
- Controla la excepción para evitar que el programa falle.

- Muestra un mensaje claro al usuario explicando la causa del error.

```python
try:
    numero = 7 / 0
except ZeroDivisionError:
    print('Error: No es posible dividir entre cero. Tiene que insertar un numero valido.')
```

2.- Segundo bloque — Índice fuera de rango

 Código a evaluar:

lista = [4, 7, 30, 23, 5]
lista[10]
                    
- Controla la excepción producida al intentar acceder a ese índice.

- Muestra un mensaje claro al usuario explicando la causa del error.

```python
lista = [4, 7, 30, 23, 5]
try:
    elemento = lista[10]
    print(elemento)
except IndexError:
    print(f"Error: El índice 10 está fuera de la lista inicial.")
```
