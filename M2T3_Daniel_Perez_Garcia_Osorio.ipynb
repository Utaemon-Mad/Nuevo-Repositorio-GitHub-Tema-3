1.- Función para calcular el área de un rectángulo

- Crea una función llamada area_rectangulo() que reciba la base y la altura como parámetros.

- Los valores de base y altura se pedirán al usuario por teclado.

- Devuelve e imprime el área resultante.
Nota: el área de un rectángulo se obtiene al multiplicar la base por la altura.

```python
def area_rectangulo(base, altura):
    return base*altura

#Funcion para pedir insertar los datos al usuario
base_usuario =float(input("Ingrese la base del rectángulo: "))
altura_usuario =float(input("Ingrese la altura del rectángulo: "))
area_usuario = area_rectangulo(base_usuario, altura_usuario)

print("El área del rectángulo es: ", area_usuario)
```

2.- Función para calcular el área de un círculo

- Crea una función llamada area_circulo() que reciba el radio como parámetro.

- El radio se pedirá al usuario por teclado.

- Devuelve e imprime el área del círculo.
Nota: el área se obtiene al elevar el radio al cuadrado y multiplicar el resultado por π (puedes usar 3.14159 o math.pi).

```python
import math

def area_circulo(radio):
    return math.pi*radio**2


radio_usuario = float(input("Inserta el radio del circulo: "))

resultado = area_circulo(radio_usuario)

print("El area del circulo es:", resultado)

```

4.- Cálculo del índice de masa corporal (IMC)

- Crea una función llamada imc() que reciba el peso (kg) y la altura (m, con dos decimales).

- Calcula el índice de masa corporal: IMC = peso / (altura * altura).

- Clasifica el resultado según la tabla:
Bajo peso: < 18.50
Normal: [18.50, 25.00]
Sobrepeso: ≥ 25.00
Obesidad: ≥ 30.00

- Muestra el resultado con su clasificación.

```python
def imc(peso, altura):
    valor_imc = peso / (altura * altura)

    if valor_imc < 18.50:
        clasificacion = "Bajo peso"
    elif valor_imc <= 25.00:
        clasificacion = "Normal"
    elif valor_imc < 30.00:
        clasificacion = "Sobrepeso"
    else:
        clasificacion = "Obesidad"

    return valor_imc, clasificacion


peso_usuario = float(input("Introduce tu peso: "))
altura_usuario = float(input("Introduce tu altura: "))

resultado_imc, categoria = imc(peso_usuario, altura_usuario)

print(f"Tu IMC es: {resultado_imc:.2f}")
print(f"Clasificación: {categoria}")
```

5.- Login estructurado en funciones

- Reestructura el programa del ejercicio 9 del Tema 2 dividiéndolo al máximo en funciones, por ejemplo: pedir_contraseña(), validar_login(), etc.

- Programa original:

Paso 1: Solicita al usuario una contraseña y guárdala en una variable.

Paso 2: Pide al usuario que introduzca la contraseña para loguearse.
Si la contraseña es correcta, muestra "Login correcto".
Si es incorrecta, muestra "Login fallido. Intento X", incrementando un contador hasta 5.
Al llegar a 5 intentos fallidos, reinicia el programa al Paso 1.


```python
MAX_INTENTOS = 5


def pedir_contraseña():
    """Solicita la contraseña inicial (Paso 1)."""
    return input("Crea tu contraseña: ").strip()


def solicitar_login():
    return input("Introduce tu contraseña para iniciar sesión: ").strip()


def validar_login(clave_correcta):
    for intento in range(1, MAX_INTENTOS):
        clave_ingresada = solicitar_login()

        if clave_ingresada == clave_correcta:
            print("Login correcto")
            return True
        else:
            print(f"Login fallido. Intento {intento}")

    return False


def iniciar_programa():
    while True:
        contraseña_guardada = pedir_contraseña()
        exito = validar_login(contraseña_guardada)

        if exito:
            break


if __name__ == "__main__":
    iniciar_programa()
```

6.- Función intermedio

- Crea una función llamada intermedio() que reciba dos números y devuelva su punto intermedio.

- Los números se deben pedir al usuario por teclado.

- Muestra el resultado en pantalla.
Nota: El punto intermedio se calcula como la suma de los dos números dividida entre 2.

Ejemplo de ejecución:

Introduce el primer número: -12
Introduce el segundo número: 24
El punto intermedio entre -12.0 y 24.0 es: 6.0

```python
def intermedio(num1, num2):
    return (num1 + num2) / 2

n1 = float(input("Introduce el primer número: "))
n2 = float(input("Introduce el segundo número: "))

resultado = intermedio(n1, n2)

print(f"El punto intermedio entre {n1} y {n2} es: {resultado}")
```

7.- Función recortar

- Crea una función llamada recortar() que reciba tres parámetros: número, límite inferior y límite superior.

- Si el número es menor que el límite inferior, devolverá el límite inferior.

- Si es mayor que el límite superior, devolverá el límite superior.

- Si está dentro del rango, se devuelve sin cambios.
Ejemplo de ejecución:

Introduce el número a recortar: 15
Introduce el límite inferior: 0
Introduce el límite superior: 10
El número recortado es: 10.0

```python
def recortar(numero, limite_inferior, limite_superior):
    if numero < limite_inferior:
        return float(limite_inferior)
    elif numero > limite_superior:
        return float(limite_superior)
    else:
        return float(numero)


num = float(input("Introduce el número a recortar: "))
lim_inf = float(input("Introduce el límite inferior: "))
lim_sup = float(input("Introduce el límite superior: "))

resultado = recortar(num, lim_inf, lim_sup)

print(f"El número recortado es: {resultado}")
```

8.- Función separar()

- Crea una función llamada separar() que reciba una lista de enteros.

- Debe devolver dos listas ordenadas: una con los números pares y otra con los impares.
Ejemplo de ejecución:

numeros = [-12, 84, 13, 20, -33, 101, 9]

Salida esperada:
[-12, 20, 84]
[-33, 9, 13, 101]

```python
def separar(lista_numeros):
    pares = []
    impares = []

    for num in lista_numeros:
        if num % 2 == 0:
            pares.append(num)
        else:
            impares.append(num)

    pares.sort()
    impares.sort()

    return pares, impares

numeros = [-12, 84, 13, 20, -33, 101, 9]


lista_pares, lista_impares = separar(numeros)

print(lista_pares)
print(lista_impares)
```
