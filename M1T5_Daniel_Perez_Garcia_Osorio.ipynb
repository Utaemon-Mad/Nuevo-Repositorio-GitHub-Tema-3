Crea una lista y una tupla (≥ 3 strings) de temática libre y realiza:

```python
# Lista y tupla

''' Una lista es una colección ordenada y modificable de elementos.
Sus características principales son:
Mantiene el orden de los elementos.
Permite elementos duplicados.
Es mutable: podemos cambiar, añadir o eliminar elementos.
Se define con corchetes: [].'''

# Podemos hacer una lista numerica, no importa si se repite algún elemento
lista_numerica = [1, 2, 3, 4, 4, 5]
print(lista_numerica)

# Tambien puede ser una lista de texto
lista_texto = ["Pepito", "Manolito", "Juanito"]
print(lista_texto)

# O incluso puede ser mixta
lista_mixta = [1, "Pepito", 3.14, True]
print(lista_mixta)

# Podemos mutar la lista, por ejemplo, cambiando un elemento
lista_numerica[0] = 10
print(lista_numerica)

# también podemos eliminar un elemento
del lista_texto[2]
print(lista_texto)
```

Crea un set (≥ 3 strings) y un diccionario (≥ 3 pares clave:valor) y realiza:

-Muestra el set y el diccionario.
-Muestra (si se puede) el segundo elemento del set y el valor del primer par del diccionario.

-Modifica (si se puede) algún elemento y muestra el resultado.

-Muestra el tamaño de ambos con len().

-Haz una búsqueda en ambos (con in para set y in/.get() para diccionario).

-Añade (si se puede) un elemento al set (.add()) y un par clave–valor al diccionario (dic["clave"] = valor).

-Borra o elimina (si se puede) el contenido de ambos (.clear()).

```python

# 1. Crear un set (>= 3 strings) y un diccionario (>= 3 pares)

elementos = {"ruedas", "alerones", "retrovisores"}

persona = {
    "nombre": "Max Verstappen",
    "edad": 28,
    "ciudad": "Belgica"
}

# 2. Mostrar el set y el diccionario

print("--- Set original ---")
print(elementos)

print("\n--- Diccionario original ---")
print(persona)


# 3. Segundo elemento del set / valor del primer par del diccionario

elementos_lista = list(elementos)
print("Set convertido a lista:", elementos_lista)
print("'Segundo' elemento:", elementos_lista[1])


primera_clave = list(persona.keys())[0]
print("Primer par del diccionario:", primera_clave, persona[primera_clave])


# 4. Modificar un elemento

elementos.remove("ruedas")
elementos.add("volante")

# En un diccionario SÍ se puede modificar directamente un valor.
persona["edad"] = 26


print("Set:", elementos)
print("Diccionario:", persona)

# 5. Tamaño de ambos con len()

print("Tamaño del set:", len(elementos))
print("Tamaño del diccionario:", len(persona))


# 6. Búsquedas

print("¿'alerones' está en el set?:", "alerones" in elementos)
print("¿'casco' está en el set?:", "casco" in elementos)

print("¿'nombre' está en el diccionario?:", "nombre" in persona)
print("Valor de 'ciudad' con .get():", persona.get("ciudad"))
print("Valor de 'profesion' con .get() (no existe):", persona.get("profesion", "No especificada"))


# 7. Añadir elementos

elementos.add("casco")
persona["profesion"] = "Piloto de F1"

print("Set:", elementos)
print("Diccionario:", persona)


# 8. Borrar el contenido de ambos

elementos.clear()
persona.clear()

print("Set:", elementos)
print("Diccionario:", persona)
```

3.-Suma de elementos en una lista

-Pide al usuario tres números float, guárdalos en una lista (sin bucles) y calcula la suma total. 

-Guarda el resultado en sumatorio.
Pista: usa sum(lista).

```python
# 1. Suma de elementos en una lista (sin bucles)

distancia1 = float(input("Introduce la distancia del circuito 1 (km): "))
distancia2 = float(input("Introduce la distancia del circuito 2 (km): "))
distancia3 = float(input("Introduce la distancia del circuito 3 (km): "))

distancias = [distancia1, distancia2, distancia3]

sumatorio = sum(distancias)

print("Lista de distancias:", distancias)
print("Suma total:", sumatorio)
```

5.-Lista de la compra

-Crea una lista con al menos 6 productos: productos = ["pan", "leche", ...].

-Muestra el segundo y el último producto.

-Modifica el tercer producto y muestra la lista actualizada.

-Intercambia el primero y el último y muestra el resultado.

-Extrae con slicing:
Tres primeros
Tres últimos
Posiciones pares
Pista: en slicing puedes definir un tercer número (salto) para tomar elementos cada n posiciones.

```python
# 1. Crear la lista con al menos 6 productos
productos = ["pan", "leche", "huevos", "arroz", "manzanas", "queso"]
print(productos)

# 2. Mostrar el segundo y el último producto
print("Segundo producto:", productos[1])
print("Último producto:", productos[-1])

# 3. Modificar el tercer producto
productos[2] = "yogur"
print(productos)

# 4. Intercambiar el primero y el último
productos[0], productos[-1] = productos[-1], productos[0]
print(productos)

# 5. Slicing
tres_primeros = productos[0:3]
tres_ultimos = productos[-3:]
posiciones_pares = productos[0::2]

print("Tres primeros:", tres_primeros)
print("Tres últimos:", tres_ultimos)
print("Posiciones pares:", posiciones_pares)
```

6.-Formateo de cadenas invertidas

-Supón que obtienes una cadena de texto corrupta (al revés). La cadena contiene nombre y año de nacimiento. Debes formatearla para obtener:

-Empleado Nombre Apellido nacido en el año Año

-Cadena de prueba: cadena = "zereP solraC, 5891"
Pista: invierte con cadena[::-1] y luego separa/da formato.

```python
cadena = "zereP solraC, 5891"

# 1. Invertir la cadena completa
cadena_invertida = cadena[::-1]
print("Cadena invertida:", cadena_invertida)

# 2. Separar por la coma (ojo: tras invertir, el año queda ANTES que el nombre)
año, nombre = cadena_invertida.split(",")

# 3. Quitar espacios sobrantes de cada parte
año = año.strip()
nombre = nombre.strip()

# 4. Formatear el resultado final
resultado = f"Empleado {nombre} nacido en el año {año}"
print(resultado)
```

7.-Operaciones con conjuntos

-Realiza las siguientes operaciones sobre dos conjuntos:

-Crea un primer conjunto llamado mamiferos con: Gato, León, Caballo, Ballena y Tortuga.

-Crea un segundo conjunto llamado peces con: Ballena y Sardina.

-Borra la Tortuga del conjunto mamiferos.

-Añade el Pez Payaso al conjunto peces.

-Imprime los conjuntos por pantalla.

-Imprime por pantalla los elementos que se encuentren en ambos conjuntos, es decir, en mamiferos y peces.


Pistas: mamiferos.remove("Tortuga") (o discard), peces.add("Pez Payaso"); intersección: mamiferos & peces o mamiferos.intersection(peces).

```python
# 1. Crear los conjuntos
mamiferos = {"Gato", "León", "Caballo", "Ballena", "Tortuga"}
peces = {"Ballena", "Sardina"}

# 2. Borrar la Tortuga de mamiferos
mamiferos.remove("Tortuga")

# 3. Añadir el Pez Payaso a peces
peces.add("Pez Payaso")

# 4. Imprimir los conjuntos
print("Mamíferos:", mamiferos)
print("Peces:", peces)

# 5. Elementos en ambos conjuntos (intersección)
comunes = mamiferos & peces
print("Elementos en ambos conjuntos:", comunes)
```

8.-Comparación de alumnos

-Crea conjuntos para alumnos de Python e IA y obtén:

-Alumnos en ambos cursos.

-Alumnos exclusivos de Python.

-Alumnos exclusivos de IA.

-Convierte cada resultado a lista y muéstralo.
Pistas: intersección &; diferencia -; conversión list(conjunto).

```python
# 1. Crear los conjuntos de alumnos
alumnos_python = {"Enzo", "Yajaira", "Daniel", "Sofía", "Adrián"}
alumnos_ia = {"Sofía", "Daniel", "Marco", "Karen"}

# 2. Alumnos en ambos cursos (intersección)
ambos_cursos = alumnos_python & alumnos_ia

# 3. Alumnos exclusivos de Python (diferencia)
solo_python = alumnos_python - alumnos_ia

# 4. Alumnos exclusivos de IA (diferencia)
solo_ia = alumnos_ia - alumnos_python

# 5. Convertir cada resultado a lista y mostrarlo
print("Alumnos en ambos cursos:", list(ambos_cursos))
print("Alumnos exclusivos de Python:", list(solo_python))
print("Alumnos exclusivos de IA:", list(solo_ia))
```

9.-Depurando duplicados

-Dada una lista con elementos repetidos:

-Genera una nueva lista con elementos únicos, eliminando repeticiones automáticamente.

-Guarda el resultado y muéstralo por pantalla.
Pistas: elimina duplicados con list(set(lista)); si quieres mantener orden, puedes usar list(dict.fromkeys(lista)).

```python
# Lista con elementos repetidos
lista = ["Ferrari", "Mercedes", "Red Bull", "Ferrari", "McLaren", "Mercedes", "Ferrari"]

print(lista)

# 1. Resultado
unicos_sin_orden = list(set(lista))
print(unicos_sin_orden)

# 2. Mateniendo el orden original
unicos_con_orden = list(dict.fromkeys(lista))
print(unicos_con_orden)
```

10.-Gestión de tareas con un diccionario

-En este ejercicio, vamos a crear un gestor de tareas donde podremos añadir información progresivamente a un diccionario.

-Contamos con un diccionario con la estructura predefinida:
"pendientes" → Lista vacía de tareas por hacer.
"en_progreso" → Lista vacía de tareas que están en marcha.
"completadas" → Lista vacía de tareas finalizadas.

-Se pedirá al usuario que introduzca el nombre de una tarea.

-Se pedirá al usuario que indique en qué estado debe añadirse: "pendientes", "en_progreso" o "completadas". La tarea se añadirá a la lista correspondiente dentro del diccionario.

-Se mostrará el diccionario actualizado con los cambios.

```python
# 1. Diccionario con la estructura predefinida
tareas = {
    "pendientes": [],
    "en_progreso": [],
    "completadas": []
}

# 2. Pedir el nombre de la tarea
nombre_tarea = input("Introduce el nombre de la tarea: ")

# 3. Pedir el estado de la tarea
estado = input("Indica el estado (pendientes / en_progreso / completadas): ")

# 4. Añadir la tarea a la lista correspondiente, comprobando que el estado exista
if estado in tareas:
    tareas[estado].append(nombre_tarea)
    # 5. Mostrar el diccionario actualizado
    print(tareas)
else:
    print(f"\nEl estado '{estado}' no es válido. Usa: pendientes, en_progreso o completadas.")
```
