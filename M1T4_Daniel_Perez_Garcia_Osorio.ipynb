2.- Redondeo: Redondea a dos decimales el resultado de: (365 / 12) * 14.7

```python
# 1. Realizamos la operación matemática
resultado = (365 / 12) * 14.7

# 2. Redondeamos a dos decimales usando round()
resultado_final = round(resultado, 2)
print(resultado_final)
```

```python
# 1. Creamos las variables con datos de prueba
username = "Admin"
password = "Tokio"

'''2. Verificación con operadores lógicos
Condición 1: El usuario debe tener entre 3 y 10 caracteres
Condición 2: La contraseña debe ser "Tokio" O "Python" '''

if (3 <= len(username) <= 10) and (password == "Tokio"):
    print("¡Acceso concedido! El usuario y la contraseña son válidos.")
else:
    print("Acceso denegado. El usuario o la contraseña no cumplen los requisitos.") 

```

Ejercicios de I/O


```python
''' 1.- Operadores de asignación: Practica con +=, -=, *=, /= 
usando dos variables numéricas. Muestra el resultado de cada paso.'''

x = 10
y = 5

# Suma
 
x += y 
print(x)

# Resta

x -= y
print(x)

# Multiplicación

x *= y
print(x)

# División

x /= y
print(x)

```

Formateo de números:
